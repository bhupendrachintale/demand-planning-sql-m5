```python
con.sql("""
SELECT item_id,
        dept_id,
        cat_id,
        total_revenue,
        ROW_NUMBER() OVER (ORDER BY total_revenue DESC, item_id) AS rank_revenue,
        SUM(total_revenue) OVER (ORDER BY total_revenue DESC,item_id ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_tot_rev,
        SUM(total_revenue) OVER () AS tot_rev_all,
        ROUND(running_tot_rev/tot_rev_all*100.0,1) AS cum_perc_rev,
        CASE WHEN cum_perc_rev <= 80 THEN 'A'WHEN cum_perc_rev <= 95 THEN 'B' ELSE 'C' END AS abc_rev
        FROM item_revenue_52w
        ORDER BY total_revenue DESC""").df()
            
```

```python
con.sql("""
WITH item_running AS (
    SELECT item_id,
           dept_id,
           cat_id,
           total_revenue,
           ROW_NUMBER() OVER (ORDER BY total_revenue DESC, item_id) AS rank_revenue,
           SUM(total_revenue) OVER (ORDER BY total_revenue DESC, item_id
                                    ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_tot_rev,
           SUM(total_revenue) OVER () AS tot_rev_all
    FROM item_revenue_52w
)
SELECT item_id,
       dept_id,
       cat_id,
       total_revenue,
       rank_revenue,
       ROUND(running_tot_rev * 100.0 / tot_rev_all, 1) AS cum_perc_rev,
       CASE WHEN running_tot_rev * 100.0 / tot_rev_all <= 80 THEN 'A'
            WHEN running_tot_rev * 100.0 / tot_rev_all <= 95 THEN 'B'
            ELSE 'C'
       END AS abc_rev
FROM item_running
ORDER BY rank_revenue
""").df()
```

```python
con.sql("""CREATE OR REPLACE TABLE item_abc AS
WITH item_running AS (
    SELECT item_id,
           dept_id,
           cat_id,
           total_revenue,
           ROW_NUMBER() OVER (ORDER BY total_revenue DESC, item_id) AS rank_revenue,
           SUM(total_revenue) OVER (ORDER BY total_revenue DESC, item_id
                                    ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_tot_rev,
           SUM(total_revenue) OVER () AS tot_rev_all
    FROM item_revenue_52w
)
SELECT item_id,
       dept_id,
       cat_id,
       total_revenue,
       rank_revenue,
       ROUND(running_tot_rev * 100.0 / tot_rev_all, 1) AS cum_perc_rev,
       CASE WHEN running_tot_rev * 100.0 / tot_rev_all <= 80 THEN 'A'
            WHEN running_tot_rev * 100.0 / tot_rev_all <= 95 THEN 'B'
            ELSE 'C'
       END AS abc_rev
FROM item_running
""")
```

```python
con.sql("""
WITH weekly_error AS (
    SELECT item_id,
           wm_yr_wk,
           total_units,
           LAG(total_units) OVER (PARTITION BY item_id ORDER BY wm_yr_wk) AS naive_fcst
    FROM item_weekly_units_52w
)
SELECT item_id,
       ROUND(STDDEV(total_units), 2) AS std_dev_demand,
       ROUND(STDDEV(naive_fcst - total_units), 2) AS std_dev_error,
       ROUND(STDDEV(naive_fcst - total_units) / STDDEV(total_units), 2) AS ratio_error_to_demand
FROM weekly_error
WHERE naive_fcst IS NOT NULL
GROUP BY item_id
ORDER BY ratio_error_to_demand DESC
""").df()
```
